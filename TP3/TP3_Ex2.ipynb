# **Verificação formal para o algoritmo extendido de Euclides**
### Trabalho prático 3 - Grupo 16

## Metodologia
Para a realização deste problema optamos por dividir-lo nas três tarefas sugeridas no enunciado.
1. Construção do CFA e traduzir para código.
2. Verificar se $\phi(a, b, r, s, t) \equiv a \times s + b \times t = r$ é um invariante, utilizando k-indução.
3. Verificar, recorrendo aos "look-aheads" que o programa termina sempre

### 1.2 Definição Formal das Transições

O autómato é constituído pelos seguintes elementos:

#### 1.2.1 Locais e Transformadores de Predicados (Weakest Pre-condition)

O conjunto de locais é $L=\{INIT, GUARD, BODY, STOP\}$.

As transições definem como o estado evolui. O transformador de predicados "weakest pre-condition" ($WP$) descreve como uma pós-condição $\psi$ é afetada pela atribuição de valores na transição. Ou seja, substituímos as variáveis pelos seus novos valores.

**Transição Inicial ($\tau_{init}$):** $INIT \to GUARD$
*   Atribuição: $r := a, \quad r' := b, \quad s := 1, \quad s' := 0, \quad t := 0, \quad t' := 1$
*   $WP(\tau_{init}, \psi) = \psi[a/r, b/r', 1/s, 0/s', 0/t, 1/t']$
    *(Isto significa: na fórmula $\psi$, substitui-se $r$ por $a$, $r'$ por $b$, etc.)*

**Transição do Corpo ($\tau_{body}$):** $BODY \to GUARD$
*   Atribuição simultânea:
    *   $q := r \div r'$
    *   $(r, r') := (r', r \pmod{r'})$
    *   $(s, s') := (s', s - q \times s')$
    *   $(t, t') := (t', t - q \times t')$
*   $WP(\tau_{body}, \psi) = \psi[r'/r, (r \pmod{r'})/r', s'/s, (s - q s')/s', t'/t, (t - q t')/t']$

**Transição de Identidade:** ($GUARD \to BODY$, $GUARD \to STOP$)
*   Estas transições não alteram variáveis, apenas mudam o local.
*   $WP(\tau_{id}, \psi) = \psi$

#### 1.2.2 Guardas (Condições de Transição)
As guardas determinam qual a aresta que pode ser percorrida a partir de um local:

*   **Em GUARD:**
    *   Para **BODY**: A condição é $r' \neq 0$.
    *   Para **STOP**: A condição é $r' = 0$.

*   **Em BODY:**
    *   Para **GUARD**: A condição é $\neg((r=0))$.

#### 1.2.3 Locais de Erro e Terminação
*   **Terminação com Sucesso:** Local **STOP**. Representa o fim da execução onde o MDC foi calculado corretamente.

```python
def extended_euclidean_trace(a, b):
    print(f"--- A calcular MDC({a}, {b}) ---")
    print(f"{'r':<5} {'rp':<5} {'s':<5} {'sp':<5} {'t':<5} {'tp':<5} | {'Verificação (as+bt=r)':<25}")
    print("-" * 70)
    
    r, rp = a, b
    s, sp = 1, 0
    t, tp = 0, 1
    
    while rp != 0:
        # Verificar invariante no início do loop
        check = (a*s + b*t == r)
        print(f"{r:<5} {rp:<5} {s:<5} {sp:<5} {t:<5} {tp:<5} | {check}")
        
        q = r // rp
        r, rp = rp, r - q * rp
        s, sp = sp, s - q * sp
        t, tp = tp, t - q * tp
        
    # Estado final
    check = (a*s + b*t == r)
    print(f"{r:<5} {rp:<5} {s:<5} {sp:<5} {t:<5} {tp:<5} | {check}")
    print("-" * 70)
    print(f"Resultado: MDC = {r}")
    print(f"Identidade de Bézout: {a}*({s}) + {b}*({t}) = {r}  -> {a*s + b*t == r}")

# Testar com valores conhecidos
extended_euclidean_trace(240, 46)
```

### 1.3 Intuição: Exemplo de Execução

Antes de passarmos à verificação formal, vamos implementar o algoritmo em Python puro para visualizar o comportamento das variáveis $(r, r', s, s', t, t')$ numa execução real. Isto ajuda a confirmar que o modelo mental do algoritmo está correto.

```python
from pysmt.shortcuts import *
from pysmt.typing import INT

def declare(i):
    state = {}
    # Usamos INT para provar a propriedade matemática sem erros de overflow
    # Isto corresponde a modelar o algoritmo num domínio ideal (Inteiros Matemáticos)
    state['a']  = Symbol(f"a_{i}", INT)
    state['b']  = Symbol(f"b_{i}", INT)
    state['r']  = Symbol(f"r_{i}", INT)
    state['rp'] = Symbol(f"rp_{i}", INT) # r'
    state['s']  = Symbol(f"s_{i}", INT)
    state['sp'] = Symbol(f"sp_{i}", INT) # s'
    state['t']  = Symbol(f"t_{i}", INT)
    state['tp'] = Symbol(f"tp_{i}", INT) # t'
    return state

def init(state):
    # Inputs > 0 e inicialização conforme o enunciado
    # Corresponde à transição INIT -> GUARD
    # r=a, r'=b, s=1, s'=0, t=0, t'=1
    return And(
        state['a'] > Int(0),
        state['b'] > Int(0),
        Equals(state['r'], state['a']),
        Equals(state['rp'], state['b']),
        Equals(state['s'], Int(1)),
        Equals(state['sp'], Int(0)),
        Equals(state['t'], Int(0)),
        Equals(state['tp'], Int(1))
    )

def trans(curr, prox):
    # Esta função modela a transição do corpo do ciclo (BODY -> GUARD)
    # O loop ocorre enquanto r' != 0 (condição de guarda em GUARD)
    cond_loop = Not(Equals(curr['rp'], Int(0)))
    
    # Cálculos Matemáticos (q = r div r')
    q = Div(curr['r'], curr['rp']) 
    
    # Atualizações conforme o código do Problema 1 e a definição de WP(tau_body):
    # r_new = r'
    # r'_new = r - q * r'
    # s_new = s'
    # s'_new = s - q * s'
    # t_new = t'
    # t'_new = t - q * t'
    
    evolution = And(
        Equals(prox['r'],  curr['rp']),
        Equals(prox['rp'], curr['r'] - (q * curr['rp'])),
        
        Equals(prox['s'],  curr['sp']),
        Equals(prox['sp'], curr['s'] - (q * curr['sp'])),
        
        Equals(prox['t'],  curr['tp']),
        Equals(prox['tp'], curr['t'] - (q * curr['tp'])),
        
        # As constantes a e b mantêm-se inalteradas
        Equals(prox['a'], curr['a']),
        Equals(prox['b'], curr['b'])
    )
    
    # A transição de estado só é válida se a guarda do loop for verdadeira
    return And(cond_loop, evolution)
```

## 2. Verificação de Correção

### 2.1 Correção Parcial: Invariante de Bézout

Nesta alínea, o objetivo é provar a correção matemática do algoritmo, verificando se a Identidade de Bézout se mantém verdadeira ao longo da execução. O invariante candidato é:$$\phi(a, b, r, s, t) \equiv a \times s + b \times t = r$$

Para verificar o invariante $\phi$, utilizaremos o método de $k$-indução. 

Ao contrário da indução simples (que olha apenas para o estado anterior), a $k$-indução permite assumir a validade da propriedade num histórico de execução de comprimento $k$.

O procedimento consiste em dois passos provados pelo SMT Solver:

- **Caso Base**: 
Verificar se $\phi$ é válido nos primeiros $k$ estados da execução a partir do estado inicial ($Init$).

- **Passo Indutivo**: 
Verificar se, assumindo que $\phi$ é válido numa sequência de $k$ transições consecutivas quaisquer, então $\phi$ também é válido no estado seguinte ($k+1$).Se ambas as condições forem satisfeitas (retornarem UNSAT para a sua negação), a propriedade é um invariante.

```python
def inv_bezout(state):
    # Invariante a provar: a*s + b*t = r
    term1 = state['a'] * state['s']
    term2 = state['b'] * state['t']
    return Equals(term1 + term2, state['r'])

def kinduction_always(declare, init, trans, inv, k):
    print(f"--- A verificar com k={k} ---")
    with Solver(name="z3") as s:
        
        # === CASO BASE ===
        # Verificar se o invariante é válido nos primeiros k estados
        states = [declare(i) for i in range(k)]
        
        # 1. Estado inicial é válido
        s.add_assertion(init(states[0]))
        
        # 2. As k-1 transições são válidas
        for i in range(k-1):
            s.add_assertion(trans(states[i], states[i+1]))
        
        # 3. Verificar se o invariante falha em ALGUM destes estados
        condition_fail = Or([Not(inv(st)) for st in states])
        
        s.push()
        s.add_assertion(condition_fail)
        if s.solve():
            print(f"[FALHA] O Caso Base falhou para k={k}.")
            print("Contra-exemplo encontrado.")
            return False
        s.pop()
        
        # === PASSO INDUTIVO ===
        # Assumir que é válido em k estados consecutivos -> provar no k+1
        states_ind = [declare(i) for i in range(k+1)]
        
        # 1. Assumir transições válidas
        for i in range(k):
            s.add_assertion(trans(states_ind[i], states_ind[i+1]))
            
        # 2. Hipótese de Indução: Invariante válido de 0 até k-1
        for i in range(k):
            s.add_assertion(inv(states_ind[i]))
            
        # 3. Tentar provar que falha no estado k
        s.add_assertion(Not(inv(states_ind[k])))
        
        if s.solve():
            print(f"[FALHA] Passo Indutivo falhou para k={k}.")
            return False
        else:
            print(f"[SUCESSO] Invariante provado por {k}-indução!")
            return True
```

```python
kinduction_always(declare, init, trans, inv_bezout, k=6)
```

### 2.2 Look Aheads

Para completar a prova de correção do programa, é necessário garantir que este termina para quaisquer entradas válidas (i.e., que não entra num ciclo infinito).
1. **Metodologia: Função Variante**

A prova de terminação baseia-se na identificação de uma Função Variante ($V$) que mapeia o estado do programa num domínio bem ordenado (neste caso, os Inteiros não-negativos). 

Para garantir a terminação, a função $V$ deve obedecer a duas propriedades:Limitação Inferior: O valor de $V$ nunca é negativo ($V(s) \ge 0$).

Decrescimento Estrito: O valor de $V$ diminui estritamente a cada $k$ passos de execução ($V(s_{k}) < V(s_{0})$).

Se uma quantidade inteira limitada inferiormente decresce sempre, ela inevitavelmente chegará a um estado onde a condição de guarda do ciclo falha (neste caso, quando atinge 0), forçando a paragem.

2. **Identificação do Variante**

No contexto do Algoritmo de Euclides, o candidato natural a variante é a variável $r'$ (rp).

A lógica do algoritmo define que, em cada iteração, o novo valor de $r'$ é o resto da divisão inteira da iteração anterior:$$r'_{new} = r \pmod{r'}$$

Pelas propriedades da aritmética modular, sabemos que para quaisquer inteiros positivos, o resto é estritamente menor que o divisor:$$0 \le (r \pmod{r'}) < r'$$

3. **Verificação com o Solver**

Utilizando a metodologia dos "look-aheads", instruímos o solver SMT a procurar um contra-exemplo: uma execução de $k$ passos onde o valor de $r'$ não decresça (ou seja, $r'_{final} \ge r'_{inicial}$) e o programa ainda não tenha terminado.Dado que a redução do resto ocorre em todas as iterações simples, basta um look-ahead de $k=1$ para provar a terminação.

```python
def variante(state):
    # O nosso variante é o valor de r' (rp). 
    # No algoritmo de Euclides, o resto diminui estritamente em cada passo.
    return state['rp']

def k_lookahead(declare, trans, var, k):
    print(f"--- A verificar terminação com lookahead={k} ---")
    with Solver(name="z3") as s:
        # Criar k+1 estados (estado 0 até estado k)
        states = [declare(i) for i in range(k+1)]
        
        # Aplicar k transições consecutivas
        for i in range(k):
            s.add_assertion(trans(states[i], states[i+1]))
            
            # Propriedade 1: O variante deve ser sempre não-negativo
            # (No nosso caso, rp >= 0 é garantido pela lógica de inteiros positivos)
            s.add_assertion(var(states[i]) >= Int(0))

        nao_decresceu = GE(var(states[k]), var(states[0]))
        nao_terminou  = Not(Equals(var(states[k]), Int(0)))
        
        # Queremos ver se é possível esta situação má acontecer
        s.add_assertion(And(nao_decresceu, nao_terminou))
        
        if s.solve():
            print(f"[FALHA] Existe um caminho onde o variante não decresce em {k} passos.")
            print("Contra-exemplo:")
            # Opcional: imprimir valores para debug
            m = s.get_model()
            rp_0 = m.get_py_value(states[0]['rp'])
            rp_k = m.get_py_value(states[k]['rp'])
            print(f"rp_inicial = {rp_0}, rp_final = {rp_k}")
            return False
        else:
            print(f"[SUCESSO] O programa termina! O variante decresce sempre a cada {k} passo(s).")
            return True
```

```python
k_lookahead(declare, trans, variante, k=500)
```

1.  **Correção Parcial (Safety):** Utilizando $k$-indução ($k=1$), provámos que a Identidade de Bézout $a \times s + b \times t = r$ é um invariante do ciclo. Isto garante que, se o programa terminar, o resultado calculado satisfaz a propriedade matemática desejada.
