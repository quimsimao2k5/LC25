# **Verificação formal para o algoritmo extendido de Euclides**
### Trabalho prático 3 - Grupo 16

## **Alínea A: Modelação SFOTS**

Nesta secção, definimos o modelo formal do Algoritmo de Euclides Estendido. O sistema é representado como um **Sistema de Transição de Primeira Ordem Simbólico (SFOTS)**, onde o estado é composto pelo contador de programa (`pc`) e pelas variáveis do algoritmo ($r, r', s, s', t, t'$).

Optámos por utilizar **BitVectors de 16 bits** para simular a aritmética de máquina. Esta escolha obriga-nos a lidar explicitamente com o problema do *overflow*, que não existe na aritmética de inteiros ideal, mas é crítico em implementações reais.

### Estados do Sistema
O sistema evolui através dos seguintes estados de controlo (`pc`):
1.  **INIT**: Estado inicial.
2.  **GUARD**: Verificação da condição de paragem ($r' \neq 0$).
3.  **BODY**: Cálculo dos novos valores e atualização das variáveis.
4.  **STOP**: Estado final de sucesso (quando $r' = 0$).
5.  **ERROR**: Estado de erro, atingido se ocorrer *overflow* nas operações aritméticas.

### Estratégia de Deteção de Overflow
Para garantir que nenhuma operação excede os limites de representação de 16 bits ($[-32768, 32767]$), implementámos uma lógica de verificação robusta:
1.  **Extensão de Sinal:** Antes de qualquer operação aritmética crítica (subtração e multiplicação), os operandos são estendidos para 32 bits.
2.  **Cálculo Seguro:** A operação é realizada neste domínio alargado, onde o *overflow* não ocorre.
3.  **Verificação de Limites:** O resultado é comparado com os valores máximo e mínimo representáveis por 16 bits.
4.  **Transição de Erro:** Se o resultado estiver fora dos limites, o sistema transita para um estado de erro absorvente (`ERROR`). Caso contrário, o valor é truncado de volta para 16 bits e o algoritmo prossegue.

```python
from z3 import *

# --- Configuração Global ---
# Definimos os estados como constantes para serem usados em todas as funções
nbits_default = 16
INIT, GUARD, BODY, STOP, ERROR = 0, 1, 2, 3, 4

def trans(curr, prox):
    pc, r, r_line = curr['pc'], curr['r'], curr['r_line']
    s, s_line = curr['s'], curr['s_line']
    t, t_line = curr['t'], curr['t_line']
    
    pc_p, r_p, r_line_p = prox['pc'], prox['r'], prox['r_line']
    s_p, s_line_p = prox['s'], prox['s_line']
    t_p, t_line_p = prox['t'], prox['t_line']

    same_vars = And(r_p == r, r_line_p == r_line, s_p == s, s_line_p == s_line, t_p == t, t_line_p == t_line)

    # 1. INIT -> GUARD
    t_init = And(pc == INIT, pc_p == GUARD, same_vars)

    # 2. GUARD -> STOP (r' == 0)
    t_guard_stop = And(pc == GUARD, r_line == 0, pc_p == STOP, same_vars)

    # 3. GUARD -> BODY (r' != 0)
    t_guard_body = And(pc == GUARD, r_line != 0, pc_p == BODY, same_vars)

    # 4. BODY -> GUARD ou ERROR
    # Cálculo q = r / r' (Divisão Unsigned porque assumimos inputs > 0)
    q = UDiv(r, r_line)
    
    # Extensão para 32 bits para verificar Overflow
    # Assumimos nbits da variável 's' (pode ser 16 ou 4, dependendo da chamada)
    # Vamos buscar o tamanho do bitvector dinamicamente
    size = s.sort().size() 
    ext_size = size * 2
    
    s_ext = SignExt(size, s)
    s_line_ext = SignExt(size, s_line)
    t_ext = SignExt(size, t)
    t_line_ext = SignExt(size, t_line)
    q_ext = ZeroExt(size, q) # q é sempre positivo
    
    # Cálculos em precisão estendida
    val_s_new = s_ext - (q_ext * s_line_ext)
    val_t_new = t_ext - (q_ext * t_line_ext)
    
    # Limites para detetar overflow
    min_val = -(2**(size - 1))
    max_val = (2**(size - 1)) - 1
    l_min = BitVecVal(min_val, ext_size)
    l_max = BitVecVal(max_val, ext_size)
    
    is_overflow = Or(
        val_s_new < l_min, val_s_new > l_max,
        val_t_new < l_min, val_t_new > l_max,
        r == 0 # Condição extra do enunciado
    )
    
    # Transição de Erro
    t_body_error = And(pc == BODY, is_overflow, pc_p == ERROR, same_vars)
    
    # Transição de Sucesso
    t_body_guard = And(
        pc == BODY,
        Not(is_overflow),
        pc_p == GUARD,
        r_p == r_line,
        r_line_p == URem(r, r_line),
        s_p == s_line,
        s_line_p == Extract(size-1, 0, val_s_new),
        t_p == t_line,
        t_line_p == Extract(size-1, 0, val_t_new)
    )

    return Or(t_init, t_guard_stop, t_guard_body, t_body_error, t_body_guard)

print("SFOTS Criado.")
```

## **Alínea B: Verificação de Invariantes (CHCs)**

Na Alínea B, procedemos à verificação formal de segurança utilizando **Constraint Horn Clauses (CHCs)**. Esta abordagem permite-nos perguntar ao solver se existe algum estado acessível que viole a propriedade de segurança (neste caso, alcançar o estado `ERROR`).

O problema é codificado através de três regras lógicas fundamentais:
1.  **Regra de Inicialização $$Init \implies Inv$$** O estado inicial do algoritmo deve satisfazer o invariante.
2.  **Regra de Transição $$Inv \land Trans \implies Inv'$$** Se o invariante é válido num estado, deve permanecer válido após qualquer transição permitida pelo sistema.
3.  **Regra de Segurança $$Inv \land Error \implies False$$** O invariante deve ser incompatível com o estado de erro. Se o solver conseguir provar isto (UNSAT), o sistema é seguro. Se encontrar uma contradição (SAT), o sistema é inseguro.

Utilizamos o motor **Spacer** do Z3, um algoritmo baseado em PDR (Property Directed Reachability), desenhado especificamente para inferir invariantes e resolver problemas de atingibilidade em sistemas infinitos ou muito grandes.

```python
from z3 import *

def verificar_seguranca_chc(nbits=16, timeout=50000):
    fp = Fixedpoint()
    fp.set(engine='spacer')
    fp.set(timeout=timeout)

    # Declarar Função Invariante e Falha
    sorts = [BitVecSort(nbits)] * 7
    Inv = Function('Inv', *sorts, BoolSort())
    Fail = Function('Fail', BoolSort())
    fp.register_relation(Inv)
    fp.register_relation(Fail)
    
    # Variáveis
    pc = BitVec('pc', nbits)
    r, r_line = BitVec('r', nbits), BitVec('r_line', nbits)
    s, s_line = BitVec('s', nbits), BitVec('s_line', nbits)
    t, t_line = BitVec('t', nbits), BitVec('t_line', nbits)
    
    # Próximos estados
    pc_p = BitVec('pc_p', nbits)
    r_p, r_line_p = BitVec('r_p', nbits), BitVec('r_line_p', nbits)
    s_p, s_line_p = BitVec('s_p', nbits), BitVec('s_line_p', nbits)
    t_p, t_line_p = BitVec('t_p', nbits), BitVec('t_line_p', nbits)

    curr = {'pc': pc, 'r': r, 'r_line': r_line, 's': s, 's_line': s_line, 't': t, 't_line': t_line}
    prox = {'pc': pc_p, 'r': r_p, 'r_line': r_line_p, 's': s_p, 's_line': s_line_p, 't': t_p, 't_line': t_line_p}
    
    vars_curr = list(curr.values())
    vars_prox = list(prox.values())

    # 1. Regra INIT
    init_cond = And(
        pc == BitVecVal(INIT, nbits),
        r > 0, r_line > 0,
        s == 1, s_line == 0,
        t == 0, t_line == 1
    )
    fp.rule(ForAll(vars_curr, Implies(init_cond, Inv(*vars_curr))))

    # 2. Regra TRANS
    fp.rule(ForAll(vars_curr + vars_prox, 
                   Implies(And(Inv(*vars_curr), trans(curr, prox)), Inv(*vars_prox))))

    # 3. Regra FAIL
    error_cond = (pc == BitVecVal(ERROR, nbits))
    fp.rule(ForAll(vars_curr, Implies(And(Inv(*vars_curr), error_cond), Fail())))
    
    print(f"A verificar segurança para nbits={nbits} (Timeout: {timeout}ms)...")
    
    try:
        result = fp.query(Fail())
    except Exception:
        result = unknown # Captura o 'canceled' e trata como unknown
    
    if result == sat:
        print("Resultado: SAT (INSEGURO - O erro é alcançável)")
    elif result == unsat:
        print("Resultado: UNSAT (SEGURO - Invariante encontrado)")
    else:
        print("Resultado: UNKNOWN (Timeout ou inconclusivo)")
        
    return result, fp

# Execução protegida da Alínea B
verificar_seguranca_chc(nbits=16, timeout=10000)
```

### Análise dos Resultados
Ao executar a verificação para 16 bits, observamos que o solver tem dificuldade em terminar ou retorna `Unknown` dentro de um tempo razoável. Isto deve-se a fatores intrínsecos ao problema:

1.  **Explosão do Espaço de Estados:** Com 16 bits, o número de estados possíveis é gigantesco ($2^{16 \times 6}$), tornando a exploração exaustiva impraticável.
2.  **Aritmética Não-Linear:** O algoritmo depende de divisões inteiras (`UDiv`) e multiplicações. Estas operações são notoriamente difíceis para solvers SMT, pois criam restrições não-convexas que dificultam a generalização de lemmas pelo motor Spacer.
3.  **Crescimento dos Coeficientes:** Os coeficientes de Bézout ($s$ e $t$) tendem a crescer rapidamente (comportamento exponencial similar à sequência de Fibonacci). Isto significa que o *overflow* é não só possível, mas provável para muitas entradas, impedindo a existência de um invariante de segurança simples que prove a ausência de erros.

## **Alínea C: Interpolantes e Traço de Erro**

Dada a dificuldade em provar a segurança (UNSAT) e a forte suspeita de que o algoritmo é inseguro, mudamos a estratégia para tentar encontrar um **contra-exemplo** (SAT).

Um contra-exemplo é um traço de execução concreto (valores iniciais e passos de transição) que começa no estado `INIT` e termina no estado `ERROR`.

Para tornar esta busca viável em tempo útil, **reduzimos a largura dos BitVectors para 4 bits**. Esta redução diminui drasticamente o espaço de busca, permitindo ao solver encontrar rapidamente uma combinação de entradas ($r, r'$) que provoca um *overflow* nos coeficientes $s$ ou $t$.

```python
print("--- Alínea C: Procura de Contra-Exemplo (Standalone) ---")

def procurar_contra_exemplo_independente():
    # Configuração específica para encontrar erro rápido
    nbits_reduzido = 4 
    timeout_ms = 5000
    
    fp = Fixedpoint()
    fp.set(engine='spacer')
    fp.set(timeout=timeout_ms)

    # 1. Redefinir Relações (Tudo local para não depender de células anteriores)
    sorts = [BitVecSort(nbits_reduzido)] * 7
    Inv = Function('Inv', *sorts, BoolSort())
    Fail = Function('Fail', BoolSort())
    fp.register_relation(Inv)
    fp.register_relation(Fail)
    
    # 2. Variáveis
    pc = BitVec('pc', nbits_reduzido)
    r, r_line = BitVec('r', nbits_reduzido), BitVec('r_line', nbits_reduzido)
    s, s_line = BitVec('s', nbits_reduzido), BitVec('s_line', nbits_reduzido)
    t, t_line = BitVec('t', nbits_reduzido), BitVec('t_line', nbits_reduzido)
    
    pc_p = BitVec('pc_p', nbits_reduzido)
    r_p, r_line_p = BitVec('r_p', nbits_reduzido), BitVec('r_line_p', nbits_reduzido)
    s_p, s_line_p = BitVec('s_p', nbits_reduzido), BitVec('s_line_p', nbits_reduzido)
    t_p, t_line_p = BitVec('t_p', nbits_reduzido), BitVec('t_line_p', nbits_reduzido)

    curr = {'pc': pc, 'r': r, 'r_line': r_line, 's': s, 's_line': s_line, 't': t, 't_line': t_line}
    prox = {'pc': pc_p, 'r': r_p, 'r_line': r_line_p, 's': s_p, 's_line': s_line_p, 't': t_p, 't_line': t_line_p}
    
    vars_curr = list(curr.values())
    vars_prox = list(prox.values())

    # 3. Regras (Init, Trans, Fail)
    # Reutilizamos a função 'trans' global que definiste na Alínea A, i
    # pois ela adapta-se dinamicamente ao tamanho dos bits das variáveis passadas.
    
    # Init
    init_cond = And(
        pc == BitVecVal(INIT, nbits_reduzido),
        r > 0, r_line > 0,
        s == 1, s_line == 0,
        t == 0, t_line == 1
    )
    fp.rule(ForAll(vars_curr, Implies(init_cond, Inv(*vars_curr))))

    # Trans
    fp.rule(ForAll(vars_curr + vars_prox, 
                   Implies(And(Inv(*vars_curr), trans(curr, prox)), Inv(*vars_prox))))

    # Fail
    error_cond = (pc == BitVecVal(ERROR, nbits_reduzido))
    fp.rule(ForAll(vars_curr, Implies(And(Inv(*vars_curr), error_cond), Fail())))

    # 4. Executar Query
    print(f"A procurar traço de erro com {nbits_reduzido} bits...")
    res = fp.query(Fail())
    
    if res == sat:
        print("\n[SUCESSO] Traço de erro encontrado!")
        print("O solver provou que o estado ERROR é acessível.")
        print("-" * 40)
        print("Detalhes do Traço (Ground Sat Answer):")
        print(fp.get_ground_sat_answer())
        print("-" * 40)
    elif res == unsat:
        print("[AVISO] O sistema parece seguro para estes bits (Inesperado para este algoritmo).")
    else:
        print("[ERRO] Timeout. Não foi possível encontrar o traço.")

# Executar a função independente
try:
    procurar_contra_exemplo_independente()
except Exception as e:
    print(f"Ocorreu um erro na execução: {e}")
```

```python
def encontrar_erro_bmc(nbits=16, max_steps=20):
    print(f"--- BMC: À procura de Overflow com {nbits} bits (até {max_steps} passos) ---")
    
    s = Solver()
    
    # 1. Inputs Simbólicos (a e b)
    a = BitVec('a_input', nbits)
    b = BitVec('b_input', nbits)
    
    # Restrição: inputs devem ser positivos
    s.add(a > 0, b > 0)
    
    # 2. Estado Inicial (Passo 0)
    # Variáveis do passo 0
    r_curr, rp_curr = a, b
    s_curr, sp_curr = BitVecVal(1, nbits), BitVecVal(0, nbits)
    t_curr, tp_curr = BitVecVal(0, nbits), BitVecVal(1, nbits)
    
    # 3. Desenrolar o Loop (Unrolling)
    for k in range(1, max_steps + 1):
        
        # --- A. Condição de Guarda ---
        # O loop só continua se r' != 0. 
        # Se r' == 0, o programa parou antes deste passo.
        cond_loop_active = (rp_curr != 0)
        
        # --- B. Simular Cálculos (Com deteção de Overflow) ---
        
        # Extensão para 32 bits (para verificar a matemática real)
        size = nbits
        ext_size = size * 2
        
        r_ext  = SignExt(size, r_curr)
        rp_ext = SignExt(size, rp_curr)
        s_ext  = SignExt(size, s_curr)
        sp_ext = SignExt(size, sp_curr)
        t_ext  = SignExt(size, t_curr)
        tp_ext = SignExt(size, tp_curr)
        
        # q = r / r'
        # Proteção contra divisão por zero simbólica no Z3
        q_ext = If(rp_ext == 0, BitVecVal(0, ext_size), r_ext / rp_ext)
        
        # Cálculos ideais em 32 bits
        # s_new = s - q * s'
        val_s_new = s_ext - (q_ext * sp_ext)
        # t_new = t - q * t'
        val_t_new = t_ext - (q_ext * tp_ext)
        
        # --- C. Verificar se houve ERRO neste passo ---
        l_min = BitVecVal(-(2**(size - 1)), ext_size)
        l_max = BitVecVal((2**(size - 1)) - 1, ext_size)
        
        overflow_s = Or(val_s_new < l_min, val_s_new > l_max)
        overflow_t = Or(val_t_new < l_min, val_t_new > l_max)
        r_is_zero  = (r_curr == 0)
        
        is_error = Or(overflow_s, overflow_t, r_is_zero)
        
        # O erro é real se:
        # 1. O loop ainda estava ativo (cond_loop_active)
        # 2. A condição de erro aconteceu (is_error)
        crash_condition = And(cond_loop_active, is_error)
        
        # --- D. Perguntar ao Z3 ---
        s.push() # Guardar contexto
        s.add(crash_condition) # Assumir que o erro acontece AGORA
        
        result = s.check()
        if result == sat:
            print(f"\n[ATENCAO] Erro encontrado no passo k={k}!")
            m = s.model()
            val_a = m[a].as_long()
            val_b = m[b].as_long()
            print(f"Inputs que levam o programa a um estado de erro:")
            print(f"  a = {val_a}")
            print(f"  b = {val_b}")
            return
            
        s.pop()
        
        next_r = rp_curr
        next_rp = SRem(r_curr, rp_curr)
        
        next_s = sp_curr
        next_sp = Extract(nbits-1, 0, val_s_new)
        
        next_t = tp_curr
        next_tp = Extract(nbits-1, 0, val_t_new)
        
        # Atualizar ponteiros para a próxima iteração
        r_curr, rp_curr = next_r, next_rp
        s_curr, sp_curr = next_s, next_sp
        t_curr, tp_curr = next_t, next_tp

    print(f"Nenhum erro encontrado até {max_steps} passos.")

# Executar BMC
encontrar_erro_bmc(nbits=16, max_steps=20)
```
